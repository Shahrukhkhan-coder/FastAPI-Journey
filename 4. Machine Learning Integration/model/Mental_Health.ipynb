```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = df.drop_duplicates()
```

```python
df['Physical_Activity_Hours'] = df['Physical_Activity_Hours'].clip(lower=0)
```

```python
df.describe()
```

```python
num_cols = df.select_dtypes(include='number')
num_cols.skew()
```

```python
df.columns
```

```python
ord_col = df.select_dtypes(include='object')
ord_col['Country'].nunique()
```

```python
counts = df['Country'].value_counts()

threshold = 100 

repl = counts[counts<=threshold].index

df['Country'] = df['Country'].replace(repl,'Other')
```

```python
Q1 = num_cols.quantile(0.25)

Q3 = num_cols.quantile(0.75)

IQR =  Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = (num_cols < lower_limit) | (num_cols > upper_limit)
outliers.sum()
```

```python
df = df[(df['Study_Hours'] < upper_limit['Study_Hours']) & (df['Study_Hours'] > lower_limit['Study_Hours'])]
```

```python
df = df[(df['Physical_Activity_Hours'] < upper_limit['Physical_Activity_Hours']) & (df['Physical_Activity_Hours'] > lower_limit['Physical_Activity_Hours'])]
```

```python
df.shape
```

```python
skew_col = ['Study_Hours']

numeric_cols = ['Age','Avg_Daily_Usage_Hours','Daily_Unlocks','Physical_Activity_Hours','Sleep_Hours_Per_Night']

ordinal_cols = ['Academic_Level','Stress_Level']

nominal_cols = ['Gender','Country','Most_Used_Platform','Purpose_Of_Use']

features = skew_col + numeric_cols + ordinal_cols + nominal_cols
x = df[features]
y = df['Mental_Health_Score']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler, FunctionTransformer, OrdinalEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

study_hours_pipeline = Pipeline(steps=[
    ('log_transform', FunctionTransformer(np.log1p)),
    ('scaler', StandardScaler())
])

preprocessing = ColumnTransformer(transformers=[
    ('study_hours', study_hours_pipeline, skew_col),
    ('numer',StandardScaler(),numeric_cols),
    ('ordinal',OrdinalEncoder(categories=[['High School', 'Undergraduate', 'Graduate'],['Low', 'Medium', 'High', 'Very High']]),ordinal_cols),
    ('nominal',OneHotEncoder(drop='first',sparse_output=False,dtype=np.int32),nominal_cols)
])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

rf_pipeline = Pipeline(steps=[
                       ('preprocessing',preprocessing),
                       ('regression',RandomForestRegressor())
])
```

```python
rf_pipeline.fit(x_train, y_train)

rf_pred_test = rf_pipeline.predict(x_test)
rf_pred_train = rf_pipeline.predict(x_train)

testing_accuracy = r2_score(y_test,rf_pred_test)
traning_accuracy = r2_score(y_train,rf_pred_train)
mae_score = mean_absolute_error(y_test, rf_pred_test)

print(f"Accuracy of Training {traning_accuracy}")
print(f"Accuracy of Testing {testing_accuracy}")
print(f'MAE {mae_score}')
```

```python
# cross validation using cross_val_score
from sklearn.model_selection import cross_val_score
cross_val_score(rf_pipeline, x_train, y_train, cv=5, scoring='r2').mean()
```

```python
param_grid = {
    'regression__n_estimators': [20, 50, 100, 150, 200],
    'regression__max_features': [0.25, 0.5, 0.75, 1.0],
    'regression__max_depth': [None, 2, 6, 10],
    'regression__max_samples': [0.5, 0.7, 0.8, 1.0]
}
```

```python
from sklearn.model_selection import GridSearchCV

rf_grid = GridSearchCV(
    estimator=rf_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring='r2',
    verbose=2,
    n_jobs=-1
)
```

```python
rf_grid.fit(x_train, y_train)
```

```python
print("Best parameters:", rf_grid.best_params_)
print("Best CV R² score:", rf_grid.best_score_)
```

```python
best_rf_model = rf_grid.best_estimator_

rf_pred_train = best_rf_model.predict(x_train)
rf_pred_test = best_rf_model.predict(x_test)

training_r2 = r2_score(y_train, rf_pred_train)
testing_r2 = r2_score(y_test, rf_pred_test)
mae = mean_absolute_error(y_test, rf_pred_test)

print(f"Training R²: {training_r2:.4f}")
print(f"Testing R²: {testing_r2:.4f}")
print(f"MAE: {mae:.4f}")
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.svm import SVR
from xgboost import XGBRegressor
from sklearn.linear_model import LinearRegression

def objective(trial):

    regressor_name = trial.suggest_categorical('regressor', ['GradientBoostingRegressor','SVR','LinearRegression','RandomForest','XGBoost'])

    if regressor_name == 'SVR':
        c = trial.suggest_float('c', 0.1, 100, log=True)
        kernel = trial.suggest_categorical('kernel', ['linear','rbf','poly','sigmoid'])
        gamma = trial.suggest_categorical('gamma', ['scale','auto'])

        model = SVR(C=c, kernel=kernel, gamma=gamma)


    elif regressor_name == 'LinearRegression':
        model = LinearRegression()


    elif regressor_name == 'GradientBoostingRegressor':
        n_estimators = trial.suggest_int('n_estimators', 50, 300)
        learning_rate = trial.suggest_float('learning_rate', 0.01, 0.3, log=True)
        max_depth = trial.suggest_int('max_depth', 3, 20)
        min_samples_split = trial.suggest_int('min_samples_split', 2, 10)
        min_samples_leaf = trial.suggest_int('min_samples_leaf', 1, 10)

        model = GradientBoostingRegressor(
            n_estimators=n_estimators,
            learning_rate=learning_rate,
            max_depth=max_depth,
            min_samples_split=min_samples_split,
            min_samples_leaf=min_samples_leaf,
            random_state=42
        )


    elif regressor_name == 'RandomForest':
        n_estimators = trial.suggest_int('n_estimators', 50, 300)
        max_depth = trial.suggest_int('max_depth', 3, 20)
        min_samples_split = trial.suggest_int('min_samples_split', 2, 10)
        min_samples_leaf = trial.suggest_int('min_samples_leaf', 1, 10)
        bootstrap = trial.suggest_categorical('bootstrap', [True, False])

        model = RandomForestRegressor(
            n_estimators=n_estimators,
            max_depth=max_depth,
            min_samples_split=min_samples_split,
            min_samples_leaf=min_samples_leaf,
            bootstrap=bootstrap,
            random_state=42
        )


    elif regressor_name == 'XGBoost':
        n_estimators = trial.suggest_int('n_estimators',50,500)
        learning_rate = trial.suggest_float('learning_rate',0.01,0.3,log=True)
        max_depth = trial.suggest_int('max_depth',2,12)
        min_child_weight = trial.suggest_int('min_child_weight',1,10)
        subsample = trial.suggest_float('subsample',0.5,1.0)
        colsample_bytree = trial.suggest_float('colsample_bytree',0.5,1.0)
        reg_alpha = trial.suggest_float('reg_alpha',1e-8,10.0,log=True)
        reg_lambda = trial.suggest_float('reg_lambda',1e-8,10.0,log=True)

        model = XGBRegressor(
            n_estimators=n_estimators,
            learning_rate=learning_rate,
            max_depth=max_depth,
            min_child_weight=min_child_weight,
            subsample=subsample,
            colsample_bytree=colsample_bytree,
            reg_alpha=reg_alpha,
            reg_lambda=reg_lambda,
            objective='reg:squarederror',
            random_state=42,
            n_jobs=-1
        )


    model_pipeline = Pipeline(steps=[
        ('preprocessing', preprocessing),
        ('regression', model)
     ])


    score =cross_val_score(model_pipeline, x_train, y_train, cv=3, scoring='r2', n_jobs=-1, error_score='raise').mean()
    return score
```

```python
import optuna

study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=100)
```

```python
print("Best trial parameters:", study.best_trial.params)
print("Best trial accuracy:", study.best_trial.value)
```

```python
#best_params = study.best_params.copy()
#best_params.pop("regressor", None)
from xgboost import XGBRegressor
best_params = {'n_estimators': 401, 'learning_rate': 0.07476401763106098, 'max_depth': 12, 'min_child_weight': 1,
               'subsample': 0.6562107391220474, 'colsample_bytree': 0.5924209511158179, 'reg_alpha': 0.009318841159975961,
               'reg_lambda': 5.032005021362352}


xgb_pipeline = Pipeline(steps=[
                       ('preprocessing',preprocessing),
                       ('regression',XGBRegressor(**best_params))
])
```

```python
xgb_pipeline.fit(x_train, y_train)

xgb_pred_test = xgb_pipeline.predict(x_test)
xgb_pred_train = xgb_pipeline.predict(x_train)

testing_accuracy = r2_score(y_test,xgb_pred_test)
traning_accuracy = r2_score(y_train,xgb_pred_train)
mae_score = mean_absolute_error(y_test, xgb_pred_test)

print(f"Accuracy of Training {traning_accuracy}")
print(f"Accuracy of Testing {testing_accuracy}")
print(f'MAE {mae_score}')
```

```python
# cross validation using cross_val_score
from sklearn.model_selection import cross_val_score
cross_val_score(xgb_pipeline, x_train, y_train, cv=5, scoring='r2').mean()
```
